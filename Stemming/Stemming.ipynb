```python
# Classification Problem
# Comments on Product is Positive or Negative Review
```

```python
# Dataset -
# Reviews (Textual Data) - contains words (eating, eat, eaten) --> "eat" or (going, go, gone, goes) -- "go"
# For above "eat" is stem of words, and having variety of words related to stem will not help in making classification for positive or negative
# Not required to have similar kind of words again and again
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
# PorterStemmer
from nltk.stem import PorterStemmer

stemming=PorterStemmer()
```

```python
for word in words:
    print(word,"-->",stemming.stem(word))
```

```python
# Stemming applied -- sometimes we dont get exact meaning of the word eg: history --> histori
```

```python
stemming.stem("congratulations") # instead of congratulate it changed to congratul (- no meaning word)
```

```python
stemming.stem("sitting") # This works fine. Stemming works for good number of words
```

```python
# For SPAM / HAM Classification
# Here we can go with Stemming
```

```python
# RegexpStemmer Class
from nltk.stem import RegexpStemmer

reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem("eating") # Why e didnt get removed from prefix, e is also mentioned in the regexp (the dollar at the end makes the exp check only for last)
```

```python
reg_stemmer.stem("ingeating")
```

```python
# Snowball Stemmer - Better Technique compared to PorterStemmer
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer(language="english")
```

```python
print(" ".join(snowball_stemmer.languages))
```

```python
for word in words:
    print(word,"-->",snowball_stemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowball_stemmer.stem("fairly"),snowball_stemmer.stem("sportingly")
```

```python
stemming.stem("goes")
```

```python
snowball_stemmer.stem("goes") # Snow Ball is better than PorterStemmer but still has come disadvantages
```
