```python
corpus="""Hello Welcome, to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
# Tokenization
# Paragraph --> Sentences
# import nltk
# nltk.download("punkt_tab")
from nltk.tokenize import sent_tokenize
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
# Tokenization
# Paragraph --> Words or Sentence --> Words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
# We get 3 lists here, the word_tokenize accepts text as parameter (Here each sentence given as parameter)
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
# Also make sure that punctuations are also treated as separate word
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
# Full Stop will not be treated as separate word. It will be included in the previous word, except the last fullstop.
tokenizer.tokenize(corpus)
```
