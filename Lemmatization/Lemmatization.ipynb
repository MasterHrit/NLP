```python
from nltk.stem import WordNetLemmatizer
# It gives the root word(valid word with same meaning), instead of root stem from the word
# It compares from dictionary, here is morphy function present in WordNet CorpusReader to find the root word
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
# import nltk
# nltk.download('wordnet')
```

```python
'''
pos:
Noun - n
Verb - v
Adjective - a
Adverb - r
'''
lemmatizer.lemmatize("going",pos="v")
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
for word in words:
    print(word,"-->",lemmatizer.lemmatize(word,pos="v"))
```

```python
lemmatizer.lemmatize("goes",pos="v")
```

```python
lemmatizer.lemmatize("fairly",pos="v"),lemmatizer.lemmatize("sportingly",pos="v")
```
